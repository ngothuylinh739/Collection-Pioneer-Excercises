## Cây nhị phân: Cây mà mỗi nút có tối đa 2 nút con ##
## Quy ước chiều cao cây nhị phân ##
- Nếu nút không trỏ đến bất cứ cây con nào --> Chiều cao bằng 0
- Nếu nút trỏ đến một nút là nullptr --> Chiều cao bằng 1.
- Vậy chiều cao của một cây là chiều cao của cây con cao nhất cộng 1.
## Yêu cầu bài toán ##
- Xác định một cây có phải là cân bằng cao hay không.
- Một cây cân bằng cao là 2 nhánh trái phải của một nút không được có chiều cao chênh lệch quá 1.
## Hướng giải ##
- Đối với cây nhị phân, muốn xét duyệt tất cả các nút trong cây đó, thì thường khi định nghĩa một hàm, sẽ lại gọi chính nó bên trong, cho đến khi gặp nút rỗng, rồi trở ra và tiếp tục xét duyệt. (Ví dụ như duyệt đến tận cùng nhánh trái tức là đã gặp nullptr của một nút nào đó, liền trở ra và duyệt nhánh bên phải của ngay nút đó)
- Vậy cần dò các nhánh con có cân bằng cao hay không, chỉ cần một nhánh con không có cân bằng cao, thì ngay lập tức toàn cây không cân bằng cao (trả về false).
- Trong một thao tác xét nút bất kỳ, cần xét xem chiều cao hai nhánh tại nút đó có cân bằng hay không.

## CODE ##


```python
/**
 * Definition for a binary tree node.
 * struct TreeNode {
 *     int val;
 *     TreeNode *left;
 *     TreeNode *right;
 *     TreeNode() : val(0), left(nullptr), right(nullptr) {}
 *     TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
 *     TreeNode(int x, TreeNode *left, TreeNode *right) : val(x), left(left), right(right) {}
 * };
 */
class Solution {
public:
    int check (TreeNode* root){
        if (root ==nullptr){return 0;}
        int left = check (root->left);
        int right =check (root->right);
        if (left ==-1|right==-1){return -1;}
        if (abs (left-right)>1){return -1;}
        return max (left,right) +1;
    }
    bool isBalanced(TreeNode* root) {
        return check (root)!=-1;
    }
};
```

## Giải thích CODE ##
- Một nút chỉ được tính chiều cao chỉ khi nút đó cân bằng cao.
- Hàm check trả về chiều cao của cây, và thêm cả tác vụ nếu không cân bằng thì trả về trừ -1.
- Một nút là rỗng thì chiều cao bằng 0 (return 0);
- Mỗi một nút được xét thì xét chiều cao trái, chiều cao phải, trong mỗi lần xét chiều cao đó, lại đi qua các nút, rồi lại tiếp tục quy trình xét cân bằng hay không, trả về chiều cao, trả về trừ 1. Khi một nút bị trả về -1, lập trức nút cha cũng sẽ là -1.
- Hàm bool isBalanced trả về kết quả true false theo đúng yêu cầu đề. Nếu -1 --> Fasle, và bất kỳ giá trị nào khác thì là true (Do phải cân bằng cao thì mới trả về được chiều cao)
- Vậy về bản chất, thì đây chính là xét từ dưới đáy xét lên.

## Độ phức tạp ##

- Độ phức tạp phụ thuộc vào số nút của cây. Ví dụ cây có n nút thì độ phức tạp là O(n).
