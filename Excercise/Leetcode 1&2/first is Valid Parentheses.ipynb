

## Yêu cầu đề bài ##
### Một chuỗi đầu vào được coi là hợp lệ nếu: ###
- Các dấu ngoặc mở phải được đóng bằng các dấu ngoặc cùng loại.
- Các dấu ngoặc mở phải được đóng theo đúng thứ tự.
- Mỗi dấu ngoặc đóng đều có một dấu ngoặc mở tương ứng cùng loại.

## Dựa vào gợi ý của đề, sử dụng stack trong C++ ##
- Xét từng ký tự của chuỗi được truyền vào, và dò xem trong phần còn lại của chuỗi có dấu nào tương ứng hay không.
- Một dấu ngoặc mở ra thì dấu ngoặc đóng lại phải tương ứng.
- Vậy thì dấu ngoặc nào mở ra đầu tiên, nó phải được đóng lại cuối cùng.
- Nhưng nếu như vậy thì các chuỗi "( ) {[]}", xét thấy chuỗi hợp lệ, nhưng không lồng nhau.
- Vì vậy về cơ bản, khi xét từng ký tự, phải xóa đi những phần đã xét rồi, để khi mở một dấu ngoặc mới, thì chỉ xét phần mở ra đó thôi.

## Chi tiết code ##


```python
class Solution {
public:
    bool isValid(string s) {
        stack <char> cpy;
        for (int i =0; i<s.size(); i++){
            if (s[i]=='['||s[i]=='{'||s[i]=='('){cpy.push(s[i]);}
            else {if (cpy.empty()){return false;}
            if ((s[i]==']'&&cpy.top()!='[')||
                (s[i]==')'&&cpy.top()!='(')||
                (s[i]=='}'&&cpy.top()!='{')){return false;} //Các điều kiện để trả về false
            cpy.pop();
            }
        }
        if (cpy.empty()){return true;} else return false;
    }
};
```

### Giải thích ###
- Hàm trả về true nếu hợp lệ và false nếu không hợp lệ.
- Mở một vòng lặp để quét tất cả các ký tự trong chuỗi truyền vào.
- Nếu ký tự được dò là ký tự mở ngoặc --> Đẩy vào stack.
- Nếu ký tự được dò là ký tự đóng ngoặc --> Dò ký tự đóng ngoặc đó, có hợp với ký tự mở cuối cùng được truyền vào stack hay không (C++ hỗ trợ, cú pháp là stack_name.top()).
- Ta không truyền ký tự đóng ngoặc vào, nếu xuất hiện tương thích --> Xóa ký tự mở tương ứng trong stack.
- Nếu ký tự đóng ngoặc không tương ứng với ký tự mở được truyền vào stack cuối cùng thì trả về false.
- Nếu chuỗi hợp lê, thì phải đến ký tự cuối cùng, phần tử mở ngoặc được đóng lại, thì stack sẽ xóa phần tử còn lại cuối cùng, thoát vòng lặp. Nên nếu stack trước khi chạy xong vòng lặp mà rỗng, điều đó có nghĩa là không có dấu ngoặc mở nào hợp lệ được mở ra, mà chuỗi vẫn còn, nên ký tự tiếp theo sẽ là đóng ngoặc. Nhưng ký tự mở ngoặc phải xuất hiện trước ký tự đóng ngoặc, nên nếu stack rỗng thì chắc chắn chuỗi không hợp lệ, trả về false.

### Độ phức tạp ###
- Độ dài của chuỗi tối đa là 10 mũ 4 phần tử.
- Trong hàm có duy nhất một vòng lặp với độ phức tạp O(n).
- Với điều kiện if- else, tất cả đều chỉ xét 1 phần tử nên độ phức tạp coi là O(1)
- Tổng kết lại thì độ phức tạp của code trên chỉ là O(n).
